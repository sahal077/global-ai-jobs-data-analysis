```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('/content/global_ai_jobs_2026.csv')
df
```

```python
df.head()
```

# Dataset Overview

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
df.info()
```

```python
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate Job IDs:", df["Job_ID"].duplicated().sum())
```

```python
# Check missing values in Required_Certifications
print("Missing certifications:",
      df["Required_Certifications"].isnull().sum())
```

```python
df["Required_Certifications"] = df["Required_Certifications"].fillna(
    "Not Specified"
)

print("Missing values after cleaning:",
      df["Required_Certifications"].isnull().sum())
```

```python
# Visualization style
sns.set_theme(
    style="darkgrid",
    context="notebook"
)

plt.rcParams["figure.facecolor"] = "#121826"
plt.rcParams["axes.facecolor"] = "#1B2433"
plt.rcParams["axes.edgecolor"] = "#64748B"
plt.rcParams["axes.labelcolor"] = "#E5E7EB"
plt.rcParams["xtick.color"] = "#CBD5E1"
plt.rcParams["ytick.color"] = "#CBD5E1"
plt.rcParams["text.color"] = "#F8FAFC"
plt.rcParams["grid.color"] = "#334155"
plt.rcParams["grid.alpha"] = 0.5

# Professional color palettes
main_colors = [
    "#38BDF8",
    "#818CF8",
    "#A78BFA",
    "#2DD4BF",
    "#FBBF24",
    "#FB7185",
    "#34D399",
    "#60A5FA"
]
```

# AI Job Demand by Specialization

```python
specialization = (
    df["AI_Specialization"]
    .value_counts()
    .sort_values()
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=specialization.values,
    y=specialization.index,
    palette=main_colors[:len(specialization)],
    hue=specialization.index,
    legend=False
)

plt.title(
    "AI Job Demand by Specialization",
    fontsize=16,
    fontweight="bold",
    pad=15
)

plt.xlabel("Number of Job Postings")
plt.ylabel("AI Specialization")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Experience_Level",
    y="Salary_USD",
    palette=["#38BDF8", "#818CF8", "#A78BFA", "#2DD4BF", "#FBBF24"],
    hue="Experience_Level",
    legend=False
)

plt.title(
    "Salary Distribution by Experience Level",
    fontsize=16,
    fontweight="bold",
    pad=15
)

plt.xlabel("Experience Level")
plt.ylabel("Salary (USD)")

plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

# AI Job Distribution by Continent

```python
continent = df["Continent"].value_counts()

plt.figure(figsize=(10, 6))

sns.barplot(
    x=continent.index,
    y=continent.values,
    palette=["#38BDF8", "#818CF8", "#A78BFA", "#2DD4BF", "#FBBF24", "#FB7185"],
    hue=continent.index,
    legend=False
)

plt.title(
    "AI Job Postings by Continent",
    fontsize=16,
    fontweight="bold",
    pad=15
)

plt.xlabel("Continent")
plt.ylabel("Number of Job Postings")

plt.xticks(rotation=25)

plt.tight_layout()
plt.show()
```

# Job Distribution by Industry


```python
industry = df["Industry"].value_counts().head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=industry.values,
    y=industry.index,
    hue=industry.index,
    palette=main_colors[:len(industry)],
    legend=False
)

plt.title("Top Industries by AI Job Postings", fontsize=16, fontweight="bold")
plt.xlabel("Number of Job Postings")
plt.ylabel("Industry")
plt.tight_layout()
plt.show()
```

# Education Requirement vs Salary

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Education_Required",
    y="Salary_USD",
    hue="Education_Required",
    palette=main_colors[:len(df["Education_Required"].unique())],
    legend=False
)

plt.title("Salary Distribution by Education Requirement",
          fontsize=16, fontweight="bold")

plt.xlabel("Education Required")
plt.ylabel("Salary (USD)")
plt.xticks(rotation=25)

plt.tight_layout()
plt.show()
```

# Automation Risk vs Job Growth

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Company_Size",
    y="Salary_USD",
    hue="Company_Size",
    palette=main_colors[:len(df["Company_Size"].unique())],
    legend=False
)

plt.title("Salary Distribution by Company Size",
          fontsize=16, fontweight="bold")

plt.xlabel("Company Size")
plt.ylabel("Salary (USD)")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

# Correlation Between Key AI Job Metrics

```python
numeric_cols = [
    "Salary_USD",
    "Job_Openings",
    "Competition_Level_Applicants",
    "Hiring_Difficulty_1_10",
    "Job_Satisfaction_1_10",
    "Work_Life_Balance_1_10",
    "AI_Investment_Company_Millions",
    "Company_AI_Maturity_1_10"
]

plt.figure(figsize=(11, 7))

correlation = df[numeric_cols].corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title(
    "Correlation Between Key AI Job Metrics",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

# AI Specialization by Job Type

```python
job_type_specialization = pd.crosstab(
    df["AI_Specialization"],
    df["Job_Type"]
)

job_type_specialization.plot(
    kind="bar",
    figsize=(11, 6),
    stacked=True,
    colormap="viridis"
)

plt.title("AI Specialization by Job Type",
          fontsize=16, fontweight="bold")

plt.xlabel("AI Specialization")
plt.ylabel("Number of Jobs")
plt.xticks(rotation=25)
plt.legend(title="Job Type")

plt.tight_layout()
plt.show()
```

#  Final AI Jobs Analysis

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 11))

# 1. Top Industries
industry = df["Industry"].value_counts().head(8)

sns.barplot(
    x=industry.values,
    y=industry.index,
    hue=industry.index,
    palette=main_colors[:len(industry)],
    legend=False,
    ax=axes[0, 0]
)

axes[0, 0].set_title("Top Industries by AI Job Postings",
                     fontsize=14, fontweight="bold")
axes[0, 0].set_xlabel("Number of Jobs")
axes[0, 0].set_ylabel("Industry")


# 2. Salary Distribution
sns.histplot(
    data=df,
    x="Salary_USD",
    bins=25,
    kde=True,
    ax=axes[0, 1]
)

axes[0, 1].set_title("AI Job Salary Distribution",
                     fontsize=14, fontweight="bold")
axes[0, 1].set_xlabel("Salary (USD)")
axes[0, 1].set_ylabel("Number of Jobs")


# 3. Job Openings vs Competition
sns.scatterplot(
    data=df,
    x="Competition_Level_Applicants",
    y="Job_Openings",
    hue="AI_Specialization",
    alpha=0.6,
    ax=axes[1, 0]
)

axes[1, 0].set_title("Job Openings vs Applicant Competition",
                     fontsize=14, fontweight="bold")
axes[1, 0].set_xlabel("Competition Level (Applicants)")
axes[1, 0].set_ylabel("Job Openings")
axes[1, 0].legend(
    title="AI Specialization",
    fontsize=8
)


# 4. Job Satisfaction vs Work-Life Balance
sns.scatterplot(
    data=df,
    x="Work_Life_Balance_1_10",
    y="Job_Satisfaction_1_10",
    hue="Job_Type",
    alpha=0.6,
    ax=axes[1, 1]
)

axes[1, 1].set_title("Job Satisfaction vs Work-Life Balance",
                     fontsize=14, fontweight="bold")
axes[1, 1].set_xlabel("Work-Life Balance (1-10)")
axes[1, 1].set_ylabel("Job Satisfaction (1-10)")
axes[1, 1].legend(
    title="Job Type",
    fontsize=8
)


plt.suptitle(
    "Global AI Jobs — Final Analysis",
    fontsize=20,
    fontweight="bold",
    y=1.02
)

plt.tight_layout()
plt.show()
```
